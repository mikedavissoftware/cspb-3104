1. (15 points)  Implement a heap

```python
# HELPER FUNCTIONS

def swap (a, j, k):
    temp = a[j]
    a[j] = a[k]
    a[k] = temp

def parent_index(j):
    return ((j + 1) // 2) - 1

def left_child_index(j):
    return ((j + 1) * 2) - 1

def right_child_index(j):
    return ((j + 1) * 2)


############################
### ASSIGNMENT FUNCTIONS ###

# Given an array a, heapify(a) will turn a into a heap
def heapify(a, size):
    position = (size - 1) // 2
    comps = 0

    while (position >= 0):
        comps += bubble_down(a, position, size - 1)
        position -= 1
    
    return comps

# Given a heap a and index j, bubble_up will bubble up the element at index j to the correct position 
def bubble_up(a, j):
    comps = 0
    if (j == 0):
        return comps
    if (a[j] < a[parent_index(j)]):
        swap(a, j, parent_index(j))
        comps += 1
        return comps + bubble_up(a, parent_index(j))

# Given a heap a and index j, bubble_down will move the element at index j down to a correct position
def bubble_down(a, j, size):
    comps = 0
    left = left_child_index(j)
    right = right_child_index(j)

    if (left > size): # if there is no left child
        return comps
    else: # if there is a left child
        if (right > size): # if there is no right child
            if (a[j] > a[left]):
                swap(a, j, left)
                comps += 1
        else: # if there is a right child as well
            target = -1
            if (a[left] < a[right]):
                target = left
                comps += 1
            else:
                target = right
            if (a[j] > a[target]):
                swap(a, j, target)
                comps += 1
                return comps + bubble_down(a, target, size)
    
    return comps

# Given a heap a, extract_min(a) will swap the root with the last child of the heap, then fix the heap
def extract_min(a, size):
    swap(a, 0, size - 1)
    return heapify(a, size - 1)
```

2) (10 points) Implement heapsort

Your function must sort the array in place, and return the number of direct comparisons (is a[i]  < a[j]?) made.
You must use your heap implementation.

```python
def heapsort(a):
  iteration = 0
  comps = 0
  comps += heapify(a, len(a))
  while (iteration < len(a)):
    comps += extract_min(a, len(a) - iteration)
    iteration += 1
  a.reverse()
  return comps
```
